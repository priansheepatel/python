```python
#PANDAS tRANSFORMATION
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.DataFrame({"id":[1,2,3,4],"name":["a","b","c","d"],"department":["it","ce","hr","hr"],"salary":[50000,20000,30000,40000],"bonus":[3000,4000,5000,6000]},index=[2,3,0,1])
df
```

```python
df2=pd.DataFrame({"id":[2,3,5],"salary":[20000,30000,40000],"city":["a1","a2","a3"]})
df2
```

```python
#APPLY()
#this method used to apply function to the data frame

df,df2
```

```python
df.apply(lambda x:x.sum() if x.dtype !='object' else x,axis=0)
```

```python
df.apply(lambda x:x.sum() if x.dtype !='object' else x,axis=1)
```

```python
df.apply(lambda row:row["salary"]+row["bonus"],axis=1)
```

```python
#RAW=TRUE/FALSE
#raw=false mean the data input in series
#raw=True mean the data input in numpy
df[["salary",'bonus']].apply(lambda row:row["salary"]+row["bonus"],axis=1,raw=False)
```

```python
df[["salary",'bonus']].apply(lambda row:row[0]+row[1],axis=1,raw=True)
```

```python
#result_type='expand'
df[["salary",'bonus']].apply(lambda row:[row["salary"]+row["bonus"],row["salary"]-row["bonus"]],axis=1,result_type='expand')
```

```python
#ARGS
df["salary"].apply(lambda x,inc:x+inc,args=(100,))
```

```python
#astype()
#it typecast given column
df.info()
```

```python
df["salary"]=df["salary"].astype(float)
df.info()
```

```python
df["bonus"]=df["bonus"].astype(float)
df.info()
```

```python
df
```

```python
#replace (oldvalue.nwvalue,(inplace=true,false))
df["salary"].replace(50000.0,53000,inplace=True)
```

```python
df
```

```python
#CONCAT
pd.concat([df,df])
```

```python
pd.concat([df,df],axis=1)
```

```python
pd.concat([df,df],axis=1,sort=True)
pd.concat([df,df],axis=0,sort=True)
```

```python
pd.concat([df,df],axis=0,sort=True,ignore_index=True)
pd.concat([df,df],axis=1,sort=True,ignore_index=True)
```

```python
#join parameter
#outer
#inner
pd.concat([df,df2],join="outer")
pd.concat([df,df2],join="inner")

pd.concat([df,df2],axis=1,join="inner")
pd.concat([df,df2],axis=0,join="inner",sort=True)
```

```python
#mergemerge two dat frame base on common keys\colums
#sql type operation
pd.merge(df,df2,how='inner',on="id")#common ids
```

```python
pd.merge(df,df2,how='left',on="id")
pd.merge(df,df2,left_on="id",right_on="id")
```

```python
pd.merge(df.set_index("id"),df2.set_index("id"),left_index=True,right_index=True)
```

```python
# suffixed=()
pd.merge(df.set_index("id"),df2.set_index("id"),left_index=True,right_index=True,suffixes=("_emp","_city"))
```

```python
#groupby
#it provide groupwise computation using custom/inbuilt function
g=df.groupby(by="department",as_index=True,sort=True)
```

```python
df
```

```python
g.groups#dictionary of coumn names as keys and index as values
```

```python
print(g["salary"].sum())#group sum of salary
print(g["bonus"].sum())#group sum of bonus
print(g["bonus"].mean())#group sum of mean
print(g["bonus"].median())#group sum of median
```

```python
#agg method
g["salary"].agg(["min","max","sum","median","mean"])
```

```python
#using with apply method
g["salary"].apply(lambda x:x.max()-x.min())
```

```python
g.sample(1)#random sample for each group
```

```python
g.size()#num of row per group
```

```python
g.nth(0)#select first record from group
```

```python
print(g["salary"].unique())#unique data in column
print(g["salary"].nunique())#number unique data in column
print(g["salary"].count())#num of none null data
```
